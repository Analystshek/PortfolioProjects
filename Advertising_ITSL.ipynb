```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_filename = '/content/drive/MyDrive/Advertising.csv'
df = pd.read_csv(data_filename, index_col=0)
df.iloc()
print(df)
```

```python
df_new = df.head(7)
print(df_new)
```

```python
plt.scatter(df_new['TV'], df_new['sales'])
plt.xlabel('TV ads')
plt.ylabel('Sales')
plt.title('TV ads vs Sales')
plt.show()
```

```python
import numpy as np
from sklearn.neighbors import KNeighborsRegressor
```

```python
plt.scatter(df['TV'], df['sales'])
plt.xlabel('TV ads')
plt.ylabel('Sales')
plt.title('TV ads vs Sales')
plt.show()
```

```python
x= df['TV'].values
y= df['sales'].values
```

```python
knn = KNeighborsRegressor(n_neighbors=1)
knn.fit(x.reshape(-1,1), y)
```

```python
print(x.reshape(-1,1))
```

```python
x_range = np.linspace(x.min(), x.max(), 200).reshape(-1,1)
y_pred = knn.predict(x_range)
```

```python
plt.scatter(x, y, label='Data')
plt.plot(x_range, y_pred, color='red', label='KNN line(k=1)')
plt.xlabel('TV ads spends')
plt.ylabel('Sales')
plt.title('KNN Reg on TV ads vs Sales')
plt.legend()
plt.show()
```

```python
knn10 = KNeighborsRegressor(n_neighbors=10)
knn10.fit(x.reshape(-1,1), y)

x_range2 = np.linspace(x.min(), x.max(), 200).reshape(-1,1)
y_pred2 = knn10.predict(x_range)

plt.scatter(x, y, label='Data')
plt.plot(x_range2, y_pred2, color='red', label='KNN line(k=10)')
plt.xlabel('TV ads spends')
plt.ylabel('Sales')
plt.title('KNN Reg on TV ads vs Sales')
plt.legend()
plt.show()
```
